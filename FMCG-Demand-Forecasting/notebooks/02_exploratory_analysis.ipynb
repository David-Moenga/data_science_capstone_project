# 02 - Exploratory Data Analysis (Summary)

## Dataset profile (from report)
- ~1.1M rows; 2021–2023; 13 stores; 102 SKUs; 7 countries; 4 channels.
- Total units sold ~65.12M; gross ~484.75M (currency units); net ~472.95M.

## Country (net sales share, absolute)
- Italy 29.01%, Spain 22.56%, Germany 18.72%, Austria 9.10%, France 9.01%, Poland 8.96%, Netherlands 2.65%.
- Note: coverage differs by country; normalize (per record/unit) for fairer comparisons.

## Categories
- Snacks 27.52% net, Beverages 22.01% (highest units), Personal Care 18.93%, Dairy 17.72%, Home Care 13.81%.
- Snacks + Beverages ~49.53% of net sales.

## Channels
- Hypermarket largest (6 stores); E-commerce (3 stores) sold more units than Supermarket (3 stores); Convenience (1 store) not directly comparable.

## Time patterns
- Weekends higher average daily demand than weekdays (see notebook 04).
- Monthly pattern: stronger May–Sep, dip, then rise toward December (suggests annual seasonality in supplied data).
