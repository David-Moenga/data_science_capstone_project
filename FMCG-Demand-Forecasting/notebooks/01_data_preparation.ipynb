# 01 - Data Preparation

## Source
- Primary file referenced in the capstone report: `fmcg_sales_3years_1M_rows.csv` (not included in this repo).
- Time span: 2021-01-01 to 2023-12-31 (~1.1M daily product/store/channel records).

## Key steps (from report)
1. Load data, parse dates, review dtypes.
2. Data quality checks:
   - No missing values, no exact duplicates in checked fields.
   - No negative `units_sold`, `list_price`, or `stock_on_hand`.
   - ~3,092 zero-sales rows retained (0.281%).
   - Date-derived fields (year/month/day/weekday) matched `date`.
   - `gross_sales = units_sold × list_price` validated; `net_sales = gross_sales × (1 − discount_pct)` validated.
   - `promo_flag` aligned with `discount_pct > 0` (discount retained as granular signal).
3. Caution: `stock_out_flag` vs `stock_on_hand` had inconsistencies noted in the report (definitions/timing unclear); both retained pending clarification.
4. No duplicate `(date, store_id, sku_id, channel)` combinations reported in checks.

## Outputs (typical)
- Cleaned/validated dataframe for downstream EDA, segmentation, and modeling.
- Aggregations prepared later (monthly SKU demand for ABC/XYZ/forecasting).
