# 05 - Feature Engineering (Forecasting)

## Target / granularity
- Monthly demand per SKU (aggregated from daily). 102 SKUs × 36 months = 3,672 SKU-month rows before lag drops.

## Features (from report)
- Calendar: `month`, `quarter`.
- Lags (prior actuals): `lag_1`, `lag_2`, `lag_3`, `lag_12` (same month previous year).
- Rolling: `rolling_mean_3` over prior 3 months (shifted, no leakage).
- Categorical: `sku_id`, `category` (one-hot encoded in modeling).

## Split
- Chronological: train 2022 (1,224 rows), test 2023 (1,224 rows); 2021 used to build lags (especially `lag_12`).
- Evaluation style described as rolling one-month-ahead in practice (uses actual prior month lags), not a single end-of-2022 12-step forecast without updates.
