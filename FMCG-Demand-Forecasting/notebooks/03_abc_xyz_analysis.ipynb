# 03 - ABC / XYZ Analysis (Summary)

## ABC (inventory consumption value)
- Value = `units_sold × purchase_cost`, summed per SKU over period.
- Cumulative thresholds (conventional): A ≤ ~80%, B ≤ ~95%, C rest.
- Results: A = 50 SKUs (49.02%, 80.04% of value), B = 29 (28.43%, 15.30%), C = 23 (22.55%, 4.66%).
- Note: A is broader than the classic “20% of SKUs” rule here; thresholds are business-tunable.

## XYZ (demand variability, monthly SKU level)
- Monthly demand aggregated by SKU across all locations/channels; built a complete SKU×month grid (36 months) so missing months count as 0.
- CV = std/mean of monthly demand. Thresholds used: X ≤ 0.5, 0.5 < Y ≤ 1.0, Z > 1.0.
- All 102 SKUs classified as **X** (avg CV ~0.107, min 0.029, max 0.242) at aggregated level → relatively stable monthly demand in aggregate; may hide store/channel volatility.

## Combined (AX/BX/CX)
- AX: 50 (80.04% value) — highest priority monitoring/replenishment control.
- BX: 29 (15.30%).
- CX: 23 (4.66%) — simpler rules possible while protecting service levels.
