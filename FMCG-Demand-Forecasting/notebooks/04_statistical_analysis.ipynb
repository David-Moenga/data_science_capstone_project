# 04 - Statistical Analysis: Weekday vs Weekend Demand

## Question
Is average daily FMCG demand different between weekdays and weekends?

- H0: mean daily demand (weekday) = mean daily demand (weekend)
- H1: means differ (α = 0.05)

## Descriptives (from report)
- Weekdays (n=781): mean ~56,276, median ~55,361, sd ~3,294
- Weekends (n=314): mean ~67,403, median ~66,505, sd ~3,850
- Weekend mean ~19.77% higher than weekday mean (average daily totals).

## Tests
- Welch independent t-test: t ≈ -45.01, p ≈ 3.37e-179 → reject H0.
- Paired weekly t-test (156 complete weeks): t ≈ -75.26, p ≈ 6.17e-124 → reject H0.

## Caveats
Statistical association only; does not prove causation or control for all time factors. Validate by store/SKU before policy changes.
